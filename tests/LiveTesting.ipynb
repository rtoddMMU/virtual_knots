```python
from vkt.core import Crossing, StrandRef, connect, disconnect
from vkt.core.virtual_link import VirtualLink

```

```python
c0 = Crossing(kind="classical", sign=1,id = 0)
c1 = Crossing(kind="classical", sign=1,id = 1)
c2 = Crossing(kind="classical", sign=1,id = 2)

print("Created crossings:")
print(c0, c1, c2)
```

```python
connect(c0.ref(0), c2.ref(1))
connect(c0.ref(1), c2.ref(0))
connect(c2.ref(0), c1.ref(1))
connect(c2.ref(1), c1.ref(0))
connect(c1.ref(0), c0.ref(1))
connect(c1.ref(1), c0.ref(0))
```

```python
tref = VirtualLink([c0,c1,c2])
```

```python
tref.components()
```
