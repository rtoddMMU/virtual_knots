```python
from vkt.core import Crossing, StrandRef,connect, disconnect
from vkt.core import parse_gauss_code, VirtualLink, VirtualLinkDiagram

```

```python
c0=Crossing(kind="classical", sign= -1)
#c1=Crossing(kind="virtual",id=1)
c2=Crossing(kind="classical",sign=-1)
```

```python
connect(c0.ref(0), c2.ref(0))
connect(c0.ref(1), c2.ref(1))
#connect(c1.ref(0), c2.ref(0))
#connect(c1.ref(1), c2.ref(1))
connect(c2.ref(0), c0.ref(1))
connect(c2.ref(1), c0.ref(0))
```

```python
virtual_tref=VirtualLinkDiagram([c0,c2])
```

```python
circles = virtual_tref.seifert_components()
    
    # Initialize: crossing_id -> [circle containing strand 0 jump, circle containing strand 1 jump]
crossing_to_circles = {c.id: [None, None] for c in virtual_tref.crossings}
    
for circle_idx, circle in enumerate(circles):
    # Walk through the circle looking for jumps
    # Circle structure: [pos, after_next, pos, after_next, ...]
    # Jump happens between consecutive positions where crossing changes
    for i in range(0, len(circle), 2):  # Step by 2 to look at positions before jump
        ref_before_jump = circle[i]
        ref_after_next = circle[(i + 1) % len(circle)]
        
        # The jump happens from ref_after_next to the next position
        # ref_after_next is at some crossing, and we jump to the other strand
        crossing_id = ref_after_next.crossing.id
        strand = ref_after_next.strand
        
        # Record which circle this jump belongs to
        crossing_to_circles[crossing_id][strand] = circle_idx
```
